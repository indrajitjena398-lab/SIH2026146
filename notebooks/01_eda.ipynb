## 1. Load Processed Transaction & Class Distribution

```python
import pandas as pd
import duckdb

con = duckdb.connect("data/bitcoin.duckdb")
df_tx = con.execute("SELECT * FROM transactions LIMIT 10").df()
print(df_tx)
```

## 2. Class Distribution Summary

```python
cls_summary = con.execute("SELECT label, COUNT(*) as cnt FROM transactions GROUP BY label").df()
print(cls_summary)
```
